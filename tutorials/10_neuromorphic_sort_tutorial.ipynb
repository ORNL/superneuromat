### Neuromorphic Sort with SuperNeuroMAT

The following tutorial, we'll create a spiking neural network that, given n integers in a list, will return their values sorted in ascending order. 

#### Planning / Logic

Our system will use the inherent time-based nature of spiking neural networks to represent our output. 

By spiking a single output neuron at different times determined by the inputted integers, our spike train will naturally sort them when we check what times the output neuron spikes.

![example spike train](SpikeTrain.png)

###### *Spike times of neuron id = 0 correspond to the delays in the network figure.*

#### Step 1: Set Up and Creating Neurons

```python
#If you haven't yet installed SuperNeuroMat, copy the following and run it in your terminal: 
#pip install superneuromat

# Let's begin by importing the SuperNeuroMAT library and naming our algorithm:
import superneuromat as snm
NSA = snm.SNN()
# NSA stands for neuromorphic sort algorithm
```

The data will be inputted as a list of n integer entries, so we'll create variables representing each of these: 

```python
#entries
nums = [8, 2, 4, 6, 3]

entries = []

#number of entries
n = 5
```

Now let's create an empty list to store input neuron id's in. We won't need one for outputs since there's only 1 designated output neuron:

```python
inputs = []
```

Let's create our single output neuron first so it'll show up first in the spike train:

```python
output = NSA.create_neuron(threshold=0)
```

The following will create n input neurons with threshold 0 and append their id's to the created list:

```python
for i in range(n):
    id = NSA.create_neuron(threshold = 0)
    inputs.append(id)
```

#### Step 2:  Creating Delayed Synapses

Next we'll create a synapse from each input neuron to the output neuron with corresponding delays from the list, nums:



```python
for i in range(n):
    NSA.create_synapse(pre_id = inputs[i], post_id = output, weight = 1, delay = nums[i])
```

#### Step 3: Creating Spikes and Simulating

Now let's add 1 spike to each input neuron at timestep 0. The delay of each individual spike will differentiate them in our spike train.

```python
for i in range(n):
        NSA.add_spike(time = 0, neuron_id = inputs[i], value = 1)
```

Lastly, we'll simulate and print our network. We need to run this network for the maximum delay plus 1 timestep for the spike's travel time.  

```python
NSA.simulate(max(nums)+1)
print(NSA)
# click "scrollable element" under the spike train to see the full spike train

# notice all input neurons (0 < id <= n) are manually prompted to spike at time step 0.
# disregard neurons with ids > n; SuperNeuroMAT uses the creation of extra neurons to simplify the application of delay. 
```

You can now experiment with inputting different numbers in the nums list and changing n accordingly to match the number of entries.

Here is our full code:

```python
import superneuromat as snm

NSA = snm.SNN()

# entries
nums = [8, 2, 4, 6, 3]

# number of entries
n=5

inputs = []

# create output neuron
output = NSA.create_neuron(threshold=0)

# create input neurons
for i in range(n):
        id = NSA.create_neuron(threshold = 0)
        inputs.append(id)

# create synapses
for i in range(n):
    NSA.create_synapse(pre_id = inputs[i], post_id = output, weight = 1, delay = nums[i])

# adding spikes
for i in range(n):
    NSA.add_spike(time = 0, neuron_id = inputs[i], value = 1)

# simulate
NSA.simulate(max(nums)+1)

# print
print(NSA)
```
