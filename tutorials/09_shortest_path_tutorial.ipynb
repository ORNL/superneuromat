# Using SuperNeuroMAT for Shortest Path Algorithms

One of the many tasks that we can utilize neural networks such as SuperNeuroMAT is a shortest path algorithm. An algorithm of this sort takes in a set of data points representing points and distances between them and outputs the shortest path between a given set of points. Shortest path programs are versatile, with uses ranging from road maps to social networks to online interactions. We can utilize SuperNeuroMAT's functions to design an algorithm for this purpose in a simplistic dataset.

For our purposes, we need a dataset that is easy to visualizes and work with. We will use a freely available dataset from Princeton consisting of only 8 points and 15 connections, corresponding to 8 neurons and 15 synapses in SuperNeuroMAT. Below is the link to the website this dataset comes from. 
<br>https://algs4.cs.princeton.edu/44sp/
<br>You can find the pre-processed text file derived from the smallest dataset on the website in the rsrc folder, which we will use for this example program. Open the file and look through the data, which consists of the id of the neuron sending the spike, the id of the neuron receiving the spike, and the distance between said neurons. Here is a diagram representing the dataset:
<br>![diagram of shortest path data set](img/shortest_path_images/shortest_path_diagram.png "shortest_path")

```python
#First, import SuperNeuroMAT
import superneuromat as snm
#As with all of the previous SuperNeuroMAT programs you have done, we start by creating a new SNN
snn = snm.SNN()
```

```python
#Next, we will take our text file of data and split it into a list
data = []
#adds the lines of the text to a list
with open("061_shortest_path_data.txt","r") as file:
    rawdata = [line.strip() for line in file]

#splits the list of lines into individual figures
for i in rawdata:
    data.extend(i.split())

#turns all the data into floats
for i in range(len(data)):
    data[i]=float(data[i])

print(data)
```

After the data is imported, we can begin creating the actual structure of our network. As there are 8 locations, we will need 8 neurons. We want them to spike whenever they receive a spike, so their thresholds are set to zero, and we only want them to spike once (so we can see the results easier), so their refractory period will be set high.

```python
#creates a blank list and adds our 8 neurons with given thresholds and refractory periods
points = []
for i in range(8):
    id = snn.create_neuron(threshold=0,refractory_period=1000000)
    points.append(id)
```

Now that we have the neurons, we need to connect them together with synapses. In the dataset, we are given 15 unique synapses that connect our 8 neurons. "a" is the sending neuron, "b" is the receiving neuron, and "c" is the distance between. We will use delay as our distance, and the time step at which each neuron spikes given an input spike at time step zero will be our shortest path time.

```python
for j in range(15):
    a=points[int(data[3*j])]
    b=points[int(data[3*j+1])]
    c=int(100*(data[3*j+2]))
    snn.create_synapse(a,b,delay=c)
```

The final step of our process will be testing the system. Choose whichever neuron you wish to spike first, and observe the spike train to see when each other neuron will spike. A datatable of the times between each point is included below.

```python
#add the spike at time step 0 to whichever neuron you choose (change neuron_id)
snn.add_spike(time=0,neuron_id=points[0],value=1)
#simulate and see the spike trains for various cases
snn.simulate(200)
snn.pretty_print(200)
```
